PASOS DE INVESTIGACIÓN

0) Investigar el funcionamiento y comportamiento de los diferentes tipos de ciberataques.

1) Definir los tipos de ataques que voy a analizar
   - Elegir 4 categorías inicialmente.
   - Incluir tráfico normal/benigno como categoría de referencia.
   - Justificar por qué se eligieron esas categorías.

2) Buscar y seleccionar datasets
   - Buscar datasets públicos que contengan esas categorías.
   - Priorizar datasets con características comparables.
   - Evaluar cantidad y calidad de los datos.

3) Examinar los datasets
   - Entender qué representa cada variable.
   - Analizar distribución de los datos.
   - Buscar valores faltantes, duplicados y anomalías.
   - Identificar diferencias entre datasets.

4) Limpiar y normalizar los datos
   - Eliminar/corregir datos problemáticos.
   - Unificar nombres y formatos.
   - Seleccionar las variables relevantes.
   - Normalizar cuando sea necesario.

5) Construir el dataset final
   - Unificar los datos si es necesario.
   - Definir las etiquetas (labels).
   - Separar características (features) y objetivo.

6) Exploración de patrones
   - Buscar características compartidas entre los distintos ataques.
   - Comparar ataques entre sí y contra tráfico benigno.
   - Determinar si existen patrones potencialmente generalizables.

7) Preparar los datos para Machine Learning
   - Dividir los datos en entrenamiento, validación y prueba.
   - Evitar que información del conjunto de prueba llegue al entrenamiento.

8) Entrenar el modelo
   - Entrenar con determinados tipos de ataques.
   - Evaluar su rendimiento sobre ataques conocidos.

9) Experimento de generalización
   - Retirar deliberadamente uno de los tipos de ataque del
     entrenamiento.
   - Entrenar el modelo con los restantes.
   - Probarlo sobre el ataque que nunca vio.

10) Evaluar resultados
   - Accuracy
   - Precision
   - Recall
   - F1-score
   - Matriz de confusión
   - Falsos positivos y falsos negativos

11) Comparar resultados
   - ¿Detecta mejor ataques conocidos que desconocidos?
   - ¿Qué tipos de ataques logra reconocer sin haberlos visto?
   - ¿Qué características parecen ser comunes?
   - ¿Dónde falla el modelo?

12) Conclusión
   - Determinar si existen patrones generalizables.
   - Determinar hasta qué punto Machine Learning puede utilizarlos
     para detectar ataques no observados.
